```python
from dotenv import load_dotenv 
import os 

load_dotenv() 
api_key = os.getenv('OPENWEATHER_API_KEY')
```

```python
import requests

url = "https://api.openweathermap.org/data/2.5/weather"
cidades = ["Sao Paulo,BR", "Valinhos,BR", "Curitiba,BR", "Jundiai,BR", "Rio de Janeiro", "Campinas,BR", "Salvador,BR", "Belém,BR", "Vinhedo,BR", "Louveira,BR"]

lista_clima = []

for cidade in cidades:
    parametros = {
        "q": cidade,
        "appid": api_key,
        "units": "metric",
        "lang": "pt_br"
    }
    resposta = requests.get(url, params=parametros)
    dados = resposta.json()

    lista_clima.append({
        "cidade": dados['name'],
        "temperatura": dados['main']['temp'],
        "sensacao_termica": dados['main']['feels_like'],
        "umidade": dados['main']['humidity'],
        "descricao": dados['weather'][0]['description']
    })
print(lista_clima)
```

```python
import pandas as pd

df_clima = pd.DataFrame(lista_clima)
df_clima
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 5))
sns.barplot(data=df_clima, x='cidade', y='temperatura', order=df_clima.sort_values('temperatura', ascending=False)['cidade'])
plt.title('Temperatura Atual por Cidade')
plt.xlabel('Cidade')
plt.ylabel('Temperatura (°C)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
